```python
import zipfile
import os

with zipfile.ZipFile("Pistols.v1-resize-416x416.yolov8.zip", 'r') as zip_ref:
    zip_ref.extractall("dataset")
```

```python
data_yaml = """
train: /content/dataset/export/images
val: /content/dataset/export/images
nc: 1
names: [ 'handgun' ]  # Replace with your class names
"""

with open('data.yaml', 'w') as f:
    f.write(data_yaml)
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
from ultralytics import YOLO
model = YOLO('yolov8n.pt')  # Load the YOLOv8n pre-trained model
model.train(data='/content/data.yaml', epochs=100)  # Adjust epochs and image size as needed
```

```python
import shutil
import os

# Define source and destination paths
source_path = '/content/runs/detect/train/weights/best.pt'  # Update 'exp' if the folder name differs
destination_path = '/content/drive/My Drive'

# Check if the file exists before copying
if os.path.exists(source_path):
    # Create destination folder if it doesn't exist
    os.makedirs(destination_path, exist_ok=True)
    shutil.copy(source_path, destination_path)
    print(f'Model saved to {destination_path}')
else:
    print(f"Error: File not found at {source_path}. Check if training completed successfully and the correct folder name is used.")
```
